```python
import pandas as pd
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
import pickle
```

```python
data = pd.DataFrame({
    'year': [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024,2025],
    'km_driven': [50000,30000,40000,500000,600000,7000,85000,45000,200000,9000,35000,10000,25000,12000,6000,15000],
    'fuel_type': [0,1,0,0,1,0,0,1,0,1,1,0,1,1,0,1],  # 0 = petrol, 1 = diesel
    'price': [2.3,3.5,1.5,6.5,4.5,5.5,5.5,3.5,2.5,3.5,1.5,2.5,1.5,3.3,4.3,2.2]
    # price in Lakhs
})
```

```python
current_year = 2025  # Replace with dynamic year if needed
data['car_age'] = current_year - data['year']
```

```python
X = data[['car_age', 'km_driven', 'fuel_type']]
y = data['price']
```

```python
model = RandomForestRegressor (n_estimators=100, random_state=42)
model.fit(X, y)
```

```python
with open('model.pkl', 'wb') as f:
    pickle.dump(model, f)
```
